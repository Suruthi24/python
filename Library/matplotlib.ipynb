```python
import matplotlib.pyplot as plt
```

```python
#Line graph

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 40]

plt.plot(x, y)
plt.show()
```

```python
import pandas as pd
titanic = pd.read_csv("train_titanic.csv")
```

```python
#scatter plot
plt.scatter(titanic["Age"], titanic["Fare"])
plt.show()
```

```python
#bar chart
plt.bar(["Male", "Female"], [577, 314])
plt.show()
```

```python
#Histogram
plt.hist(titanic["Age"].dropna())
plt.show()
```

```python
#Pie chart
gender_count = titanic["Sex"].value_counts()

plt.pie(
    gender_count.values,
    labels=gender_count.index
)

plt.show()
```

```python
#Box plot
plt.boxplot(titanic["Age"].dropna())
plt.show()
```

```python
#labeling
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.plot([1, 2, 3], [10, 20, 30], label="Sales")

plt.legend()
plt.show()  
plt.grid()
plt.xlim(0, 80)
plt.ylim(0, 500)
plt.xticks([0, 20, 40, 60, 80])
plt.figure(figsize=(10, 6))

plt.hist(titanic["Age"].dropna())

plt.show()
```
